```python
import nltk

sentence="ai is simulation of human intelligence by machine.it includes expect system NLP speech recognition and machine vision"
tokens=nltk.word_tokenize(sentence)
bigrams=list(nltk.bigrams(tokens))
print(bigrams)
```

```python
import nltk

sentence="ai is simulation of human intelligence by machine.it includes expect system NLP speech recognition and machine vision"
tokens=nltk.word_tokenize(sentence)
trigrams=list(nltk.trigrams(tokens))
print(trigrams)
```
