```python
import spacy
from nltk import ngrams
nlp=spacy.load('en_core_web_sm')
doc=nlp("ai is simulation of human intelligence by machine it includes expect system nlp speech recognization and machine vision")
bigrams=ngrams(doc,2)
print(list(bigrams))
```

```python
import spacy
from nltk import ngrams
nlp=spacy.load('en_core_web_sm')
doc=nlp("ai is simulation of human intelligence by machine it includes expect system nlp speech recognization and machine vision")
trigrams=ngrams(doc,3)
print(list(trigrams))
```
